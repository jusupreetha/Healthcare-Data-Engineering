```python
from pyspark.sql import functions as F

base_path = "file:/Workspace/Users/supreethaudayakumar@gmail.com/"

tables = [
    "appointments",
    "appointment_diagnoses",
    "appointment_treatments",
    "billing",
    "departments",
    "diagnoses",
    "doctor_departments",
    "doctors",
    "insurance_claims",
    "insurance_providers",
    "lab_test_results",
    "lab_tests",
    "medications",
    "patient_addresses",
    "patient_contacts",
    "patient_insurance",
    "patients",
    "payments",
    "prescription_items",
    "prescriptions",
    "treatments"
]

dataframes = {}

for table in tables:
    df = (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(f"{base_path}{table}.csv")
    )
    
    dataframes[table] = df
    print(f"{table}: {df.count()} rows")
```

```python
for table, df in dataframes.items():
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(f"bronze_{table}")
    )
    
    print(f"Created Bronze table: bronze_{table}")
```
